SmartMart plans to offer category-wise discounts:

```python
#Question 60: Profit After Discount
#Assuming costs remain unchanged, calculate the new profit after the discounts are applied.
discount_amount = sales_data * discount_rates
net_sales_value=sales_data-discount_amount
new_profit=net_sales_value-cost_data
print(new_profit)
```

```python
#Question 61: Profit Margin After Discount
#Calculate the new profit margin after applying discounts.
discount_amount = sales_data * discount_rates
net_sales_value = sales_data - discount_amount
new_profit = net_sales_value - cost_data
new_margin = (new_profit / net_sales_value) * 100
print(np.round(new_margin, 2))
```

```python
#Question 62: Loss-Making Transactions
#Check whether any store-category combination becomes loss-making after discounts.
#Display the affected store and category names.
discount_amount=sales_data*discount_rates
net_sales_values=sales_data-discount_amount
loss=np.where(net_sales_values < cost_data)
if loss[0].size > 0:
  print('Loss-making transactions found:')
  for i,j in zip(loss[0],loss[1]):
    print(f"Store: {stores[i]}, Category: {categories[j]}, Net Sales: {net_sales_values[i,j]:.2f}, Cost: {cost_data[i,j]:.2f}")
else:
  print('No loss-making transactions found after discounts.')
```

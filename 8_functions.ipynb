```python
#Assignment 3: Calculator Using Functions
#Create four separate functions:
#add()
#subtract()
#multiply()
#divide()
def add(number1, number2):
    return number1 + number2

def subtract(number1, number2):
    return number1 - number2

def multiply(number1, number2):
    return number1 * number2

def divide(number1, number2):
    if number2 == 0:
        return "Cannot divide by zero"
    return number1 / number2

number1 = float(input("Enter first number: "))
number2 = float(input("Enter second number: "))
operation = input("Enter operation: ")

if operation == "add":
    result = add(number1, number2)
elif operation == "subtract":
    result = subtract(number1, number2)
elif operation == "multiply":
    result = multiply(number1, number2)
elif operation == "divide":
    result = divide(number1, number2)
else:
    result = "Invalid operation"

print("Result:", result)
```

```python
#Assignment 31: Return Multiple Values
#Write a function named number_statistics()
def number_statistics(numbers):
    minimum = min(numbers)
    maximum = max(numbers)
    total = sum(numbers)
    average = total / len(numbers)
    return minimum, maximum, total, average

numbers = list(map(int, input("Enter numbers: ").split()))

minimum, maximum, total, average = number_statistics(numbers)

print("Minimum:", minimum)
print("Maximum:", maximum)
print("Total:", total)
print("Average:", average)

```

```python
#Assignment 34: Prime Number Function
#Write a function named is_prime().
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
print("Prime numbers from 1 to 100:")
for i in range(1, 101):
    if is_prime(i):
        print(i, end=" ")
```

```python
#Assignment 35: Fibonacci Function
#Write a function named generate_fibonacci().
def generate_fibnocci(terms):
    if terms <= 0:
        return []
    if terms == 1:
        return [0]
    series = [0, 1]
    while len(series) < terms:
        series.append(series[-1] + series[-2])
    return series
n = int(input("Enter number of terms: "))
print(generate_fibnocci(n))
```

```python
#Assignment 75: Filter Dates by Year Given dates in DD-MM-YYYY format:
from datetime import datetime

dates = [
    "15-08-2024",
    "20-01-2025",
    "31-12-2024",
    "10-05-2023",
    "invalid-date",
    "25-06-2025",
]
year = input("Enter year: ")


def is_valid_and_matches(date_str):

    parts = date_str.split("-")
    if (
        len(parts) == 3
        and all(p.isdigit() for p in parts)
        and len(parts[2]) == 4
    ):
        d, m, y = map(int, parts)
        if 1 <= m <= 12 and 1 <= d <= 31:

            return datetime.strptime(date_str, "%d-%m-%Y").year == int(year)
    return False


result = list(filter(is_valid_and_matches, dates))
print(result)
```

```python
#Assignment 92: Closure-Based Bank Balance Write a function named create_bank_account().
#Requirements Store the balance in the outer function. Define inner functions for: Deposit Withdrawal Balance enquiry Use nonlocal to update the balance. Reject withdrawals greater than the available balance. Return all three inner functions.
def create_bank_account(balance):
  def deposit(amount):
    nonlocal balance
    balance=balance+amount
    print("deposited",amount)
  def withdrawl(amount):
    nonlocal balance
    if amount<=balance:
      balance-=amount
      print("withdrawl",amount)
    else:
      print("insufficient balance")
  def check_balance():
    print("balance",balance)
  return deposit,withdrawl,check_balance
deposit,withdrawl,check_balance=create_bank_account(1000)
check_balance()
deposit(500)
check_balance()
withdrawl(300)
check_balance()
withdrawl(1500)
check_balance()
```

```python
#Assignment 93: Function Decorator Create a decorator named execution_logger().
#Requirements The decorator should print a message before the original function executes. It should print another message after execution. Apply it to an addition function. Preserve the original function's return value.
def execution_logger(func):
  def inner(*args,**kwargs):
    print("starting function")
    result=func(*args,**kwargs)
    print("function completed")
    return result
  return inner
@execution_logger
def add_numbers(a,b):
  result=a+b
  print("result:",result)
  return result
add_numbers(188,89)
```

```python
def it_emps(employees):
  it_sector=list(filter(lambda x:x["department"]=="IT",employees))
  return it_sector
result=it_emps(employees)
print(result)
```

```python
def calculate_revised_salaries(emp_list):
    def update_salary(emp):
        updated_emp = emp.copy()
        score = emp["performance_score"]
        if score >= 90:
            hike = 0.15
        elif score >= 80:
            hike = 0.10
        elif score >= 70:
            hike = 0.05
        else:
            hike = 0.00
        updated_emp["revised_salary"] = round(emp["salary"] * (1 + hike), 2)
        return updated_emp
    revised_list = list(map(update_salary, emp_list))
    return revised_list
revised_employees = calculate_revised_salaries(employees)
print("--- Revised Salaries ---")
for emp in revised_employees:
    print(
        f"Name: {emp['name']:<8} | Original Salary: ₹{emp['salary']} | Revised Salary: ₹{emp['revised_salary']}"
    )
```

```python
from functools import reduce
def calculate_total_payroll(emp_list):
  total_payroll=reduce(lambda x,y:x + y["salary"],emp_list,0)
  return total_payroll
total_salary=calculate_total_payroll(employees)
print("total_salary",total_salary)
```

```python
from functools import reduce
def average_score(emp_list):
  if not emp_list:
    return 0.0
  total_score=reduce(lambda x,y:x + y["performance_score"],emp_list,0)
  average=total_score/len(emp_list)
  return round(average,2)
result=average_score(employees)
print(result)
```

```python
def categorize_experience(emp_list):
  def get_band(emp):
    updated_emp=emp.copy()
    exp=emp["experience"]
    if exp >=8:
      band="senior"
    elif exp>=4:
      band="mid-level"
    else:
      band="junior"
    updated_emp["experience_band"]=band
    return updated_emp
  return list(map(get_band,emp_list))
categorized_employees=categorize_experience(employees)
for emp in categorized_employees:
    print(
        f"Name: {emp['name']:<8} | Experience: {emp['experience']} yrs | Band: {emp['experience_band']}"
    )
```

```python
from functools import reduce
def department_salary(emp_list):
  departments=set(emp["department"] for emp in emp_list)
  dept_salaries={}
  for dept in departments:
    dept_emps=filter(lambda emp:emp["department"]== dept,emp_list)
    total_dept_salary=reduce(lambda x,y:x +y["salary"],dept_emps,0)
    dept_salaries[dept]=total_dept_salary
  return dept_salaries
result=department_salary(employees)
for dept,total in result.items():
  print(f"Department: {dept:<10} | Total Salary: ₹{total:,}")
```
